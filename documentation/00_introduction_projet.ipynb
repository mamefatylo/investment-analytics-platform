# 1. Présentation du projet

Cette note ouvre le projet. Elle fixe le cadre : à quoi sert la plateforme, pour quel client elle est écrite, quelles capacités elle démontre, et comment chaque phase est conduite. Le détail de chaque étape est écrit dans la note de cette étape.

## 1.1 Contexte

Les banques privées, les sociétés de gestion d'actifs et les équipes de gestion des risques s'appuient sur des données financières pour construire des portefeuilles, mesurer les risques, analyser les performances et accompagner la décision d'investissement.

Ces données viennent de plusieurs éditeurs. Le travail consiste à les :

- collecter ;
- contrôler ;
- enrichir, c'est-à-dire leur ajouter une information qui se déduit d'une source déjà détenue ;
- analyser ;
- transformer en indicateurs, c'est-à-dire en chiffres établis toujours de la même façon, afin que deux dates se comparent ;
- présenter dans des tableaux de bord et des rapports.

## 1.2 La plateforme

Le projet développe une plateforme de Portfolio Construction, Monitoring & Risk Analytics fondée sur un mandat discrétionnaire fictif libellé en francs suisses (CHF).

Le client fixe l'objectif et les limites. Le choix des titres se fait à l'intérieur de ces limites.

Le mandat vise les actions internationales cotées. L'alignement sur les Objectifs de développement durable (ODD), les dix-sept objectifs adoptés par les Nations unies, est une condition d'entrée du mandat. L'analyse financière demeure le principal critère de décision d'investissement.

Le projet est un démonstrateur. CLIENT001 est un cas d'école. Aucun compte n'est ouvert dans une banque. Aucun ordre d'achat ou de vente n'est envoyé à une bourse. Les chiffres sont calculés sur des cours publics.

## 1.3 Métiers visés

| Fonction | Travail exercé |
|---|---|
| Portfolio Analyst | Construire un portefeuille, le suivre, expliquer son résultat |
| Investment Data Analyst | Rassembler les données, en contrôler la qualité, en documenter l'origine |
| Risk Analyst | Mesurer le risque, lire les expositions, suivre les limites du mandat |
| Wealth Management Analyst | Traduire la situation du client en contraintes d'investissement |
| Investment Operations Analyst | Tenir les référentiels, et faire en sorte que chaque chiffre puisse être refait |

## 1.4 Ordre du travail

Le travail part du client et s'achève par le rapport remis au lecteur.

```text
Profil d'investisseur
        ↓
Politique d'investissement (IPS)
        ↓
Méthodologie d'éligibilité ODD
        ↓
Univers investissable
        ↓
Acquisition des données
        ↓
Contrôle qualité
        ↓
Construction des portefeuilles
        ↓
Suivi des portefeuilles
        ↓
Analyse de performance
        ↓
Analyse du risque
        ↓
Tests de résistance
        ↓
Tableaux de bord
        ↓
Reporting
```

Le profil décrit le client : son objectif, son horizon, les pertes qu'il accepte. La politique d'investissement, ou IPS, traduit ce profil en règles. La méthodologie d'éligibilité ODD fixe les conditions auxquelles une société peut entrer dans la liste. L'univers investissable est cette liste. L'acquisition rassemble les données auprès de leurs éditeurs. Le contrôle qualité signale les manques et les valeurs qui ne peuvent pas servir. La construction fixe la part de chaque titre. Le suivi décrit la situation du portefeuille. L'analyse de performance mesure le résultat. L'analyse du risque mesure les pertes. Les tests de résistance appliquent un choc défini à l'avance. Les tableaux de bord montrent la situation sur une page. Le reporting est le rapport remis au lecteur.

## 2.1 Alignement ODD

L'alignement sur les ODD est la condition d'entrée nommée à la section 1. La satisfaire n'augmente pas la part d'un titre. L'analyse financière reste le critère de la décision. Un titre difficile à vendre, ou un risque jugé trop élevé, reste jugé sur ce terrain.

Aucun objectif n'est prioritaire par rapport aux autres. En couvrir plusieurs ne change pas la part d'un titre.

Le mandat ne présente pas la détention d'une action comme la réalisation de l'objectif : il n'y a pas d'impact direct. Il ne présente pas le placement comme la cause d'un résultat qui n'aurait pas existé sans lui : il n'y a pas d'additionnalité.

1. Pas de pauvreté
2. Faim « zéro »
3. Bonne santé et bien-être
4. Éducation de qualité
5. Égalité entre les sexes
6. Eau propre et assainissement
7. Énergie propre et d'un coût abordable
8. Travail décent et croissance économique
9. Industrie, innovation et infrastructure
10. Inégalités réduites
11. Villes et communautés durables
12. Consommation et production responsables
13. Mesures relatives à la lutte contre les changements climatiques
14. Vie aquatique
15. Vie terrestre
16. Paix, justice et institutions efficaces
17. Partenariats pour la réalisation des objectifs

## 2.2 Interdits

Le profil interdit :

- les armes controversées ;
- la vente à découvert, soit la vente d'un titre que l'on ne détient pas ;
- le levier, soit l'emprunt destiné à investir au-delà du capital du mandat ;
- les fonds cotés conçus pour amplifier un indice, ou pour évoluer en sens inverse de cet indice ;
- le capital-investissement et les autres titres non cotés ;
- les produits structurés complexes.

Le portefeuille est limité à des actifs cotés, liquides, valorisés de manière fiable, et compatibles avec une gestion internationale en actions.

La politique nomme la dette privée, prêt qui n'est pas un titre coté, parmi les instruments interdits. Elle autorise, en plus des actions internationales cotées, les fonds actions cotés.

# 4. Méthodologie

Chaque phase produit quatre composantes complémentaires.

**Réflexion métier.** La phase dit pourquoi elle existe : le problème qu'elle traite, le lecteur auquel elle répond, et le lien avec les métiers de la section 1.

**Livrable.** La phase produit un résultat qu'un responsable, un client ou un utilisateur peut ouvrir. Selon la phase, ce résultat est une analyse, un schéma, une base de données, un enchaînement de traitements, un portefeuille, un tableau de bord ou un rapport.

**Documentation.** La phase écrit les objectifs, les hypothèses, les données utilisées, les traitements et les décisions. Un tiers peut refaire le travail à partir de ce texte.

**Démonstration.** La phase apporte la preuve que le résultat tient : un exemple, un extrait de données, un contrôle, un carnet de calcul, une visualisation ou un test.

Une phase est close lorsque cette preuve est au dossier.

Cette note est la phase 0. La réflexion métier est le choix de partir du client. Le livrable est cette note. La documentation est le profil, l'ordre des règles et l'identification des documents du 16 octobre 2026. La démonstration est la conformité de ce profil à ces documents.

Le classement numéroté des notes est le planning du cadrage, `documentation/01_cadrage_projet.ipynb`.


# 5. Dictionnaire

**Action cotée.** Part d'une société achetée et vendue en bourse, avec un cours publié.

**Banque privée.** Banque qui gère le patrimoine de personnes, client par client, selon un mandat écrit.

**Capital-investissement.** Part dans une société qui n'est pas en bourse.

**Exposition.** Répartition du portefeuille entre des titres, des secteurs, des zones ou des devises.

**Gestion des risques.** Fonction qui mesure les pertes et suit les limites.

**Produit structuré.** Contrat dont le gain dépend d'une formule.

**Référentiel.** Liste de référence tenue de façon qu'un chiffre puisse être refait.

**Société de gestion d'actifs.** Société dont le métier est d'investir l'argent de ses clients selon une politique écrite.

**Titre non coté.** Titre sans cours public.

**Traçabilité.** Possibilité de retrouver l'origine d'un chiffre et la règle qui l'a produit.
