# Méthodologie d'éligibilité ODD

**Client :** CLIENT001
**Version :** 1.1
**Date :** 16 octobre 2026


## 1. Objet

La méthodologie ODD définit les règles permettant de déterminer si une entreprise est éligible aux Objectifs de développement durable (ODD) dans le cadre du mandat.

Elle répond exclusivement à la question : cette entreprise satisfait-elle les conditions d'éligibilité ODD du mandat ?

Elle ne décide pas :

- de l'achat d'un titre ;
- du poids dans le portefeuille ;
- de l'allocation ;
- du rendement attendu ;
- du risque financier ;
- de l'optimisation.

Ces décisions relèvent de l'IPS et de la phase quantitative.


## 2. Principe fondamental

L'éligibilité ODD est un filtre préalable à l'investissement.

Une entreprise doit franchir successivement les étapes suivantes :

Activités économiques, contribution ODD identifiable, preuves, contrôle de non-préjudice, analyse des controverses, contrôle qualité des données, statut d'éligibilité.

Une entreprise n'est jamais considérée éligible simplement parce qu'elle :

- possède un bon score ESG ;
- publie un rapport durable ;
- communique sur les ODD ;
- appartient à un fonds durable ;
- figure dans le fonds iShares MSCI Global Sustainable Development Goals ETF.

L'éligibilité doit être documentée, démontrable, traçable et reproductible.


## 3. Référentiel utilisé

La méthodologie utilise comme référence les 17 Objectifs de développement durable des Nations unies (ONU).

Principes :

- aucun ODD n'est prioritaire ;
- une entreprise peut être associée à un ou plusieurs ODD ;
- une entreprise peut ne correspondre à aucun ODD ;
- le nombre d'ODD associés n'influence pas le poids futur dans le portefeuille.


## 4. Unité d'analyse

L'entreprise est l'unité d'investissement. L'activité économique est l'unité de preuve.

L'analyse est réalisée selon la logique : ODD, cible, activité économique, produit ou service, entreprise. Non l'inverse.

La méthodologie interdit d'attribuer un ODD sur la seule base d'une description générale ou marketing de l'entreprise.


## 5. Définition d'une contribution ODD

Une contribution est reconnue lorsqu'une activité, un produit ou un service peut être relié de manière explicite, économiquement compréhensible, documentée, vérifiable et reproductible à un ODD ou à une cible ODD.

L'activité réelle prime toujours sur l'intention déclarée.

Ne constituent pas des preuves suffisantes :

- engagements futurs ;
- promesses ;
- objectifs internes ;
- campagnes marketing ;
- simple mention des ODD dans un rapport.


## 6. Sources d'information

**Niveau 1, priorité maximale**

- Rapports annuels
- Rapports financiers
- Rapports de durabilité
- Documents réglementaires
- Publications officielles de l'entreprise

**Niveau 2**

- Sources publiques et institutionnelles
- Autorités et organismes réglementaires

**Niveau 3**

- Fournisseurs ESG
- Fournisseurs financiers
- Bases spécialisées

**Niveau 4**

- Sources secondaires utilisées uniquement pour compléter ou corroborer une analyse

Une source secondaire ne remplace jamais une preuve primaire lorsqu'elle est disponible.


## 7. Hiérarchie des preuves

**Preuve forte.** Lien direct et clairement démontré entre l'activité et l'ODD.

**Preuve modérée.** Lien plausible mais comportant une part d'incertitude.

**Preuve faible.** Lien fondé principalement sur le marketing, la communication, la notation ESG ou une catégorisation indirecte.

Une preuve faible seule est insuffisante.


## 8. Activités mixtes

Une entreprise peut exercer simultanément des activités contribuant aux ODD, des activités neutres, des activités controversées et des activités incompatibles.

La méthodologie documente les activités concernées, conserve les proportions lorsqu'elles sont disponibles, et n'invente jamais de pourcentage de chiffre d'affaires.

L'existence d'une petite activité positive ne suffit pas à considérer l'ensemble de l'entreprise comme aligné ODD.


## 9. Seuil de contribution

Aucun seuil universel de chiffre d'affaires n'est imposé dans la version 1.1.

La méthodologie refuse :

- les seuils arbitraires ;
- les seuils inventés pour agrandir l'univers ;
- les seuils inventés pour réduire artificiellement l'univers.

Les futurs seuils éventuels devront être documentés, justifiés et versionnés.


## 10. Contrôle de non-préjudice

Une contribution positive ne suffit pas à elle seule.

Une entreprise doit également être analysée quant à :

- ses activités controversées ;
- ses violations graves ;
- sa compatibilité avec les exclusions du mandat ;
- les conflits éventuels avec les ODD attribués.

Une activité positive ne compense pas automatiquement une activité gravement problématique. Il n'existe aucun système de score net dans lequel une contribution positive moins un impact négatif rendrait l'entreprise éligible.


## 11. Controverses

Chaque controverse est évaluée selon sa nature, sa gravité, sa matérialité, sa persistance, sa crédibilité, la réponse de l'entreprise et son statut actuel.

| Niveau | Traitement |
|---|---|
| Mineure | Surveillance |
| Significative | Analyse renforcée |
| Majeure | Non-éligibilité ou statut indéterminé |
| Violation grave et persistante | Exclusion potentielle |


## 12. Gestion des données

**Données manquantes.** L'absence de preuve n'est pas une preuve d'éligibilité. Si l'information essentielle manque, le statut est Indeterminate.

**Informations contradictoires.** Lorsqu'une contradiction importante ne peut être résolue, le statut est Indeterminate.

**Données obsolètes.** Chaque décision doit comporter une date d'information, une date d'évaluation et la version méthodologique. L'éligibilité est toujours temporelle.


## 13. Statuts officiels

| Statut | Signification |
|---|---|
| Eligible | Conditions satisfaites |
| Not_Eligible | Conditions non satisfaites |
| Indeterminate | Informations insuffisantes ou contradictoires |
| Excluded | Titre interdit par le mandat |

Un statut Indeterminate n'est jamais considéré comme éligible par défaut.


## 14. Relation avec l'IPS

L'univers final résulte de l'intersection de l'univers de marché, de l'IPS et de la méthodologie ODD.

- Une entreprise ODD-éligible peut rester non investissable.
- Une entreprise financièrement attractive mais non éligible ODD reste exclue.

L'éligibilité ODD est nécessaire mais jamais suffisante.


## 15. Relation avec le benchmark

La référence stratégique est le MSCI ACWI Sustainable Development Index (Net). Net signifie : dividendes après impôt. C'est l'indice, la liste de règles à laquelle le mandat se compare. On ne l'achète pas. Il n'a pas de cours propre.

Le proxy investissable est l'iShares MSCI Global Sustainable Development Goals ETF (SDG). C'est le fonds coté qui suit cet indice. Il a un prix, une feuille de positions et un rendement publié. La comparaison publiée utilise ce fonds.

La présence d'une entreprise dans ce proxy n'est pas une preuve d'éligibilité ODD. L'absence d'une entreprise de ce proxy n'est pas une preuve de non-éligibilité.

La référence stratégique, le proxy investissable et la méthodologie sont totalement indépendants.


## 16. Qualité des données et audit

Chaque décision doit être traçable, documentée, datée, reproductible et versionnée.

Pour chaque classification, il doit être possible d'identifier :

- l'entreprise ;
- l'ODD attribué ;
- l'activité concernée ;
- les preuves utilisées ;
- les controverses identifiées ;
- le raisonnement appliqué ;
- la version de méthodologie.


## 17. Gouvernance

Sont interdits :

- la création d'exceptions pour améliorer un rendement ;
- la modification des règles pour améliorer un backtest ;
- l'invention de données ;
- l'invention de seuils ;
- l'invention de pourcentages de chiffre d'affaires ;
- le rétro-ajustement des classifications.

Toute modification importante entraîne une nouvelle version méthodologique.


## 18. Limites reconnues

La méthodologie ne mesure pas :

- l'impact réel du portefeuille ;
- l'additionnalité ;
- la causalité ;
- la durabilité absolue ;
- la qualité ESG globale ;
- la performance environnementale ou sociale totale.

Elle mesure uniquement l'éligibilité ODD d'une entreprise selon des règles documentées.


## 19. Distinction fondamentale

| Concept | Signification |
|---|---|
| ESG | Caractéristiques environnementales, sociales et de gouvernance |
| Durabilité | Concept général |
| Alignement ODD | Contribution documentée à un ODD |
| Impact | Effet dans l'économie réelle |
| Additionnalité | Effet qui n'aurait pas eu lieu autrement |

Le mandat revendique un alignement ODD documenté. Il ne revendique pas automatiquement un impact direct ni une additionnalité mesurable.


## Conclusion

La méthodologie ODD v1.1 établit un cadre rigoureux, documenté et reproductible permettant de déterminer l'éligibilité ODD des entreprises à partir de leurs activités économiques réelles. L'éligibilité repose sur des preuves vérifiables, un contrôle de non-préjudice, l'analyse des controverses et la qualité des données. Elle est indépendante de la référence stratégique, du proxy investissable, de l'optimisation financière et de la construction du portefeuille. Les règles méthodologiques sont désormais verrouillées et constituent la référence officielle pour filtrer l'univers investissable avant toute analyse financière ou quantitative.


## Dictionnaire

Ce dictionnaire reprend les termes employés dans cette note qui n'y sont pas déjà expliqués. Il n'en ajoute pas.

**Backtest.** Calcul mené sur des cours déjà passés. Cette méthodologie interdit d'en changer les règles pour améliorer le résultat.

**Cible ODD.** Sous-objectif officiel rattaché à l'un des 17 Objectifs de développement durable.

**Phase quantitative.** Choix des poids, du risque et du rendement. Cette méthodologie ne les décide pas.
