# Investment Policy Statement (IPS)

**Client :** CLIENT001
**Mandat :** gestion discrétionnaire
**Montant sous gestion :** CHF 1 000 000
**Devise de référence :** CHF
**Date :** 16 octobre 2026



## Objectif

Constituer un portefeuille international d'actions cotées visant la croissance du capital à long terme, dans le respect des contraintes de risque, de diversification, de liquidité et d'éligibilité ODD.


## Profil du mandat

- Profil : Dynamique
- Horizon minimum : 7 ans
- Horizon cible : 10 ans
- Besoin de liquidité : faible
- Dépendance aux revenus ou plus-values : aucune
- Exposition internationale : autorisée
- Couverture de change : non systématique


## Univers autorisé

**Instruments autorisés**

- Actions internationales cotées
- ETF actions cotés
- Combinaison actions/ETF

**Instruments interdits**

- Vente à découvert
- Levier financier
- ETF inverses
- ETF à levier
- Private equity
- Dette privée
- Actifs non cotés
- Produits structurés complexes
- Instruments insuffisamment liquides
- Instruments sans valorisation fiable


## Politique ODD

L'éligibilité ODD constitue une condition obligatoire d'entrée dans l'univers investissable.

L'éligibilité ODD :

- ne garantit pas une pondération supérieure ;
- ne remplace pas l'analyse financière ;
- ne compense pas une mauvaise liquidité ou un risque excessif.


## Benchmark

**Référence stratégique.** MSCI ACWI Sustainable Development Index (Net). Net signifie : dividendes après impôt. C'est l'indice, la liste de règles à laquelle le mandat se compare. On ne l'achète pas. Il n'a pas de cours propre.

**Proxy investissable.** iShares MSCI Global Sustainable Development Goals ETF (SDG). C'est le fonds coté qui suit cet indice. Il a un prix, une feuille de positions et un rendement publié. La comparaison publiée utilise ce fonds.

Le benchmark est une référence de comparaison et non une obligation de réplication.


## Contraintes de construction

| Paramètre | Limite |
|---|---|
| Positions économiques | 25 à 40 |
| Exposition maximale par émetteur | 5 % |
| Exposition maximale par secteur | 20 % |
| Exposition maximale par grande région | 50 % |
| Exposition maximale par devise | 40 % |
| Trésorerie opérationnelle | ≤ 2 % |
| Tracking error ex ante cible | ≤ 8 % |
| Turnover annuel brut cible | ≤ 50 % |


## Liquidité

- Les titres doivent être suffisamment négociables.
- Une position individuelle ne doit normalement pas dépasser 10 % de la valeur quotidienne moyenne négociée sur 20 séances.
- Les titres présentant une liquidité insuffisante ou une valorisation incertaine sont exclus.

## Dictionnaire

Ce dictionnaire reprend les termes employés dans cette note qui n'y sont pas déjà expliqués. Il n'en ajoute pas.

**CVaR.** Perte moyenne dans les pires cas. Le suivi minimal le prévoit.

**Dette privée.** Prêt qui n'est pas un titre coté. Interdite.

**ETF.** Fonds coté. Les ETF actions cotés sont autorisés. Les ETF à levier ou inversés sont interdits.

**Gestion discrétionnaire.** Le client fixe l'objectif et les limites. Le choix des titres se fait à l'intérieur de ces limites.

**Greenwashing.** Présenter un placement comme plus durable que les preuves ne le permettent.

**Levier.** Emprunt destiné à investir au-delà du capital. Interdit.

**Look-ahead bias.** Utiliser, à une date, une information qui n'était pas encore connue.

**Long-only.** Poids positifs ou nuls.

**Maximum drawdown.** Plus forte baisse depuis un sommet.

**Objectifs de développement durable (ODD).** Les 17 objectifs des Nations unies. L'éligibilité ODD est une condition d'entrée.

**Overfitting.** Ajuster les règles au passé pour améliorer un résultat historique.

**Private equity.** Part dans une société qui n'est pas en bourse. Interdit.

**Produit structuré complexe.** Contrat dont le gain dépend d'une formule. Interdit.

**Stop-loss.** Vente automatique après une baisse fixée. Cette politique n'en impose pas.

**Tracking error ex ante.** Distance prévue avec le proxy investissable, avant d'investir. Plafond : 8 %.

**Turnover.** Somme des changements de poids sur l'année. Plafond : 50 %.

**Vente à découvert.** Vente d'un titre que l'on ne détient pas. Interdite.

**Volatilité annualisée.** Ampleur des variations, ramenée à une année.
