# Règles des portefeuilles — mandat CLIENT001

Ce carnet fixe les règles de construction, de suivi et de performance des trois portefeuilles. La construction est la phase 5, le suivi la phase 6, la performance la phase 7. Il appartient à la phase 0 et se lit avec l'introduction. Les termes sont définis à l'endroit où ils apparaissent.

Les documents du 16 octobre 2026 sont dans ce dossier : le profil du client (`01_profil_investissement.ipynb`), la politique d'investissement (`02_investment_policy_statement.ipynb`) et la méthodologie d'éligibilité ODD, version 1.1 (`03_methodologie_eligibilite_odd.ipynb`). Là où le profil laissait un point ouvert et où la politique d'investissement le tranche, c'est la politique qui s'applique.

La devise de référence est le franc suisse. Les trois portefeuilles partagent le même univers, la liste des titres, et la même série de rendements en francs. Ce rendement est la variation du cours de clôture converti en francs. Le dividende versé par la société n'est pas ajouté.


## Le client

Mandat discrétionnaire : le client fixe l'objectif et les limites, et le choix des titres se fait à l'intérieur de ces limites. Profil Dynamique : croissance du capital sur un horizon long, avec une perte temporaire acceptée, sans vente à découvert, sans levier et sans titre non coté.

| Dette | aucune |
| Dépendance au revenu ou aux plus-values du portefeuille | aucune |
| Objectif | croissance du capital à long terme. Cet objectif est le premier. |

| Propension au risque | moyenne à élevée. Baisse que le client accepte de voir. Elle se distingue de la capacité. |

La performance financière reste le premier objectif. L'alignement sur les Objectifs de développement durable (ODD) est une condition d'entrée du mandat. Il n'augmente pas un poids, ne remplace pas l'analyse financière et ne compense ni une liquidité insuffisante ni un risque excessif. Aucun objectif n'est prioritaire. Le nombre d'objectifs couverts ne change pas le poids. Le mandat ne présente pas la détention d'une action comme la réalisation de l'objectif : il n'y a pas d'impact direct. Il ne présente pas le placement comme la cause d'un résultat qui n'aurait pas existé sans lui : il n'y a pas d'additionnalité.

Le client accepte une exposition en devises étrangères. Le profil laisse la couverture de change ouverte. La couverture de change est un contrat qui annule l'effet du cours d'une monnaie. La politique d'investissement la tranche : elle n'est pas systématique. La conversion en francs, sans ce contrat, est donc conforme.

Le profil interdit la vente à découvert, soit la vente d'un titre que l'on ne détient pas ; le levier, soit l'emprunt destiné à investir au-delà du capital ; les fonds cotés conçus pour amplifier un indice ou pour évoluer en sens inverse ; le capital-investissement, part dans une société qui n'est pas en bourse, et les autres titres non cotés, titres sans cours public ; les produits structurés complexes, contrats dont le gain dépend d'une formule ; les armes controversées. La politique d'investissement nomme en outre la dette privée, prêt qui n'est pas un titre coté, et elle écarte les instruments difficiles à vendre ou dont le prix n'est pas fiable.


## Limites des portefeuilles

La politique d'investissement autorise les actions internationales cotées, les fonds actions cotés, et la combinaison des deux. La référence stratégique est le MSCI ACWI Sustainable Development Index (Net). Net signifie : dividendes après impôt. C'est l'indice, la liste de règles à laquelle le mandat se compare. On ne l'achète pas. Il n'a pas de cours propre. Le proxy investissable est le fonds iShares MSCI Global Sustainable Development Goals, fonds coté dont le sigle est ETF. C'est le fonds qui suit cet indice. Il a un prix, une feuille de positions et un rendement publié. La comparaison publiée utilise ce fonds : sa feuille fournit le secteur et le pays, et son rendement publié, converti en francs, sert de comparaison. Cette feuille ne contient que des actions. Les trois livres publiés sont donc des livres d'actions. Les plafonds ci-dessous sont ceux de la politique, et les données permettent de les tenir. Le poids est la part du portefeuille dans un titre.

Le plafond de 5 % s'applique à chaque ligne de cotation. La politique appelle cela un plafond par émetteur. Deux cotations de la même société sont deux lignes, et chacune est plafonnée à 5 %.

Les zones sont au nombre de quatre : Amériques, Europe, Asie-Pacifique, Moyen-Orient et Afrique. Elles viennent du pays inscrit sur la feuille de positions. Un pays absent de cette liste est rangé dans « Non classé », et ce groupe reste plafonné à 50 %. Le secteur est le métier indiqué sur cette feuille, dans la langue de cette feuille. Un secteur absent est aussi rangé dans « Non classé ». La devise de cotation est la monnaie du cours en bourse.

Le rendement qui sert à juger ces poids est la variation du cours de clôture en francs. Le dividende, somme versée à l'actionnaire, n'entre pas dans ce chiffre. Le fonds iShares est jugé sur le rendement qu'il publie, converti en francs. Ce n'est pas le même objet.

| Règle | Application |

| Secteur | 20 % au plus |

| Rotation brute | Somme des changements de poids : 12,5 % au plus à chaque rééquilibrage, soit 50 % sur l'année |

Cette bande est imposée à deux livres. La volatilité minimale cherche le livre qui a le moins varié sur les 60 mois connus. Le Sharpe maximal cherche le plus de rendement par unité de variation sur ces mêmes mois. La variance est l'ampleur des variations d'un titre avant la racine qui donne la volatilité. Le solveur cherche ces poids. S'il dépasse 40 lignes, les 40 plus gros poids sont gardés. S'il en a moins de 25, des titres sont ajoutés selon la préférence du livre : variance la plus basse pour la volatilité minimale, excédent de rendement moyen le plus haut pour le Sharpe maximal. Les poids sont ensuite reprojetés.
