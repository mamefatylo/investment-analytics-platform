# Profil d'investissement du client fictif

**Client :** CLIENT001
**Version :** 1.4
**Date :** 16 octobre 2026
**Type de mandat :** gestion discrétionnaire axée sur les ODD (ONU)


## 1. Situation patrimoniale

- Patrimoine financier total : CHF 5 000 000
- Montant confié au mandat : CHF 1 000 000 (20 % du patrimoine)
- Avoirs hors mandat : CHF 4 000 000
- Réserve de liquidité hors mandat : CHF 500 000
- Aucun endettement
- Aucune dépendance aux revenus ou plus-values du portefeuille


## 2. Objectif d'investissement

Le mandat vise la croissance du capital à long terme et la réalisation d'un rendement financier compétitif, au travers d'une exposition internationale aux actions cotées répondant à des critères d'éligibilité ODD. La performance financière demeure l'objectif principal.


## 3. Horizon et liquidité

- Horizon minimum : 7 ans
- Horizon cible : 10 ans
- Aucun retrait important prévu durant les cinq premières années
- Besoin de liquidité : faible
- Existence d'une réserve de liquidité externe suffisante


## 4. Profil de risque

**Capacité de risque.** Élevée, notamment parce que :

- le mandat représente seulement 20 % du patrimoine ;
- le client dispose d'actifs importants hors mandat ;
- aucune dette ni dépendance financière au portefeuille n'existe.

**Propension au risque.** Moyenne à élevée :

- acceptation des fluctuations de marché ;
- tolérance à des pertes temporaires pouvant dépasser 30 % ;
- absence de réaction automatique de vente en période de baisse.

**Classification globale.** Profil Dynamique (non spéculatif).


## 5. Préférences de durabilité (ODD)

- L'alignement aux 17 Objectifs de développement durable des Nations unies est une condition préalable d'éligibilité.
- Aucun ODD n'est prioritaire à ce stade.
- Le filtre ODD ne remplace pas l'analyse financière.
- Le mandat ne revendique aucun impact direct ni aucune additionnalité mesurable.


## 6. Contraintes et exclusions

Interdiction de :

- armes controversées ;
- vente à découvert ;
- effet de levier ;
- ETF à effet de levier ou inverses ;
- private equity et autres actifs non cotés ;
- produits structurés complexes.

Exclusions encore à arbitrer dans l'IPS :

- tabac ;
- armes conventionnelles ;
- charbon thermique ;
- pétrole et gaz ;
- nucléaire ;
- jeux d'argent ;
- alcool.


## 7. Univers d'investissement autorisé

Le portefeuille doit être composé d'actifs :

- cotés ;
- liquides ;
- valorisables de manière fiable ;
- compatibles avec une gestion internationale en actions.

Le client accepte les expositions aux devises étrangères, tandis que la politique de couverture de change reste à définir.


## 8. Connaissances du client

Le client possède :

- des connaissances financières générales intermédiaires ;
- une expérience des actions, obligations, fonds et ETF ;
- une compréhension des risques de marché et de change ;
- aucune volonté d'utiliser des produits complexes ou à levier.


## 9. Décisions transférées à l'IPS

L'étape suivante devra définir notamment :

1. le benchmark ;
2. l'allocation stratégique d'actifs (SAA) ;
3. les limites de concentration ;
4. les règles de liquidité ;
5. la politique de change ;
6. les seuils d'exclusion sectorielle ;
7. les mesures de risque ;
8. les règles de rééquilibrage ;
9. la structure finale du reporting et des fichiers de suivi.


## Conclusion

CLIENT001 présente une capacité de risque élevée, une propension au risque moyenne à élevée et un horizon d'investissement de long terme. Le mandat vise la croissance du capital au travers d'actions internationales cotées répondant aux critères d'éligibilité ODD.

Les paramètres opérationnels de gestion, les limites d'investissement et les règles de portefeuille sont définis dans `02_investment_policy_statement.ipynb`.


## Dictionnaire

Ce dictionnaire reprend les termes employés dans cette note qui n'y sont pas déjà expliqués. Il n'en ajoute pas.

**Additionnalité.** Effet qui n'aurait pas eu lieu sans le placement. Le mandat ne la revendique pas.

**Allocation stratégique d'actifs (SAA).** Répartition de long terme entre les grandes catégories de placement. Ce profil la laisse à la politique d'investissement.

**Arme controversée.** Arme interdite dès ce profil.

**Benchmark.** Référence de comparaison du portefeuille. Ce profil ne la choisit pas.

**Couverture de change.** Contrat qui annule l'effet du cours d'une monnaie. Ce profil ne la fixe pas.

**Effet de levier.** Emprunt destiné à investir au-delà du capital. Interdit.

**ETF.** Fonds coté. Ce profil interdit celui qui amplifie un indice, ou qui évolue en sens inverse de cet indice.

**Gestion discrétionnaire.** Le client fixe l'objectif et les limites. Le choix des titres se fait à l'intérieur de ces limites.

**Impact direct.** Idée que détenir l'action réalise l'objectif. Le mandat ne le revendique pas.

**IPS (Investment Policy Statement).** La politique d'investissement. Elle tranche les points que ce profil laisse ouverts.

**Private equity.** Part dans une société qui n'est pas en bourse. Interdit, avec les autres actifs non cotés.

**Produit structuré complexe.** Contrat dont le gain dépend d'une formule. Interdit.

**Vente à découvert.** Vente d'un titre que l'on ne détient pas. Interdite.
