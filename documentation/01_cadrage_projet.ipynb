
# Phase 1 – Cadrage du projet
Cette phase de cadrage définit le contexte, les objectifs, le périmètre, les parties prenantes, les métiers ciblés et la feuille de route globale du projet.

## 1.1 Cahier de cadrage
### 1.1.1 Contexte et enjeux
Dans le cadre de ma transition vers les métiers de l’analyse d’investissement, de la gestion de portefeuille et de la gestion des risques en banque privée, ce projet consiste à développer une plateforme complète de **Portfolio Construction, Monitoring & Risk Analytics** appliquée à un univers d’investissement aligné sur les Objectifs de Développement Durable (ODD) des Nations Unies. Le projet vise à reproduire le fonctionnement d’une équipe Investment Analytics en couvrant l’ensemble du cycle de traitement de l’information financière, depuis la définition de l’univers investissable jusqu’au reporting et à l’aide à la décision. 
Les institutions financières s’appuient quotidiennement sur des données de marché afin de construire des portefeuilles, suivre leurs performances, mesurer les risques et fournir aux décideurs des informations fiables. La qualité des données, la rigueur analytique et la pertinence des indicateurs jouent un rôle central dans ces activités. 


### 1.1.2 Problématique métier
Comment construire une plateforme permettant :

- de définir un univers d’investissement aligné sur les ODD ; 
- d’acquérir et de fiabiliser des données financières ; 
- de construire et optimiser des portefeuilles ; 
- de mesurer la performance et le risque ; 
- de produire des outils d’aide à la décision destinés aux professionnels de l’investissement 

### 1.1.3 Objectifs du projet
Le projet vise à :

- Collecter et structurer des données financières à partir de sources publiques. 
- Garantir la qualité et l’intégrité des données utilisées. 
- Construire plusieurs stratégies de portefeuille selon différentes approches quantitatives. 
- Mesurer la performance et les risques associés. 
- Réaliser des analyses de sensibilité et des stress tests. 
- Produire des tableaux de bord et des reportings d’investissement. 
- Démontrer des compétences directement transférables vers les métiers de Portfolio Analyst, Investment Data Analyst, Risk Analyst et Wealth Management Analyst.
