## 1. Cahier de cadrage

### 1.1 Pourquoi cette phase

Cette phase fixe le cadre. Le calcul vient dans les phases suivantes. Avant de produire un chiffre, le projet écrit le contexte, le besoin, les utilisateurs et les objectifs.

### 1.2 Contexte

Le projet développe une plateforme de construction, de suivi et d'analyse du risque pour le mandat CLIENT001. Le client fixe l'objectif et les limites. Le choix des titres se fait à l'intérieur de ces limites.

CLIENT001 est un cas d'école. Aucun compte n'est ouvert dans une banque. Aucun ordre d'achat ou de vente n'est envoyé à une bourse. Les chiffres sont calculés sur des cours publics, le prix publié d'un titre ou d'une monnaie. Le résultat est jugé en francs suisses.

Le mandat vise les actions internationales cotées. L'alignement sur les Objectifs de développement durable (ODD), les dix-sept objectifs adoptés par les Nations unies, est une condition d'entrée du mandat. L'analyse financière reste le critère de la décision. Le profil, la politique d'investissement, la méthodologie d'éligibilité et les règles des portefeuilles sont les documents de la phase 0.


### 1.3 Besoin

Le travail part du client et s'achève par un rapport. Il faut des données publiques, une règle pour chaque chiffre, et une source pour chaque donnée.

Les poids ont besoin des cours, du change, du volume de chaque séance, du registre d'éligibilité ODD, et des sources des exclusions validées : tabac, armes conventionnelles, charbon thermique, jeux d'argent. Ces données entrent dans les poids quand l'acquisition les a. Une donnée manquante reste écrite comme manquante.

### 1.4 Utilisateurs

Les utilisateurs sont CLIENT001, la direction qui lit le rapport pour décider, et les métiers nommés dans l'introduction : Portfolio Analyst, Investment Data Analyst, Risk Analyst, Wealth Management Analyst, Investment Operations Analyst.

### 1.5 Objectifs

- Tenir le mandat écrit en phase 0.
- Acquérir et contrôler les données publiques dont les poids ont besoin.
- Construire trois portefeuilles sur le même univers et la même série de rendements en francs. L'équipondéré part de parts égales. La volatilité minimale est le portefeuille le moins variable. Le Sharpe maximal cherche le plus de rendement par unité de variation.
- Mesurer la performance, les pertes et les chocs écrits à l'avance.
- Publier le suivi, les tableaux de bord et le rapport.
- Exercer le travail des cinq métiers nommés au paragraphe 1.4.


### 1.6 Périmètre inclus

- le mandat déjà écrit en phase 0 ;
- les cours, le change, le volume, le registre ODD et les sources d'exclusion du tabac, des armes conventionnelles, du charbon thermique et des jeux d'argent ;
- les contrôles de complétude, de cohérence, de validité et d'unicité ;
- les trois portefeuilles, leur suivi et leur performance ;
- l'analyse du risque, les tests de résistance, les tableaux de bord et le reporting, chacun dans sa phase.

### 1.7 Hors périmètre

- l'envoi d'un ordre à une bourse ;
- l'ouverture ou la tenue d'un compte réel ;
- les contrôles d'identité du client et la lutte contre le blanchiment ;
- l'envoi automatique d'ordres ;
- l'usage dans le système vivant d'une banque ;
- la gestion d'argent réel.


### 1.8 Contraintes

Le projet est mené par Faty Barbosa Monteiro.

- Les données sont publiques.
- Le résultat est jugé en francs.
- Un tiers peut refaire un chiffre à partir des fichiers et des notes.
- L'origine de chaque chiffre et la règle qui l'a produit restent retrouvables. C'est la traçabilité.
- Les fichiers reçus restent séparés des fichiers transformés.
- Aucun statut, aucun seuil et aucun historique n'est inventé.

Les outils de ce projet sont Python, pandas, NumPy, SciPy, Plotly, SQLite et Git. Les tableaux de bord publiés sont des pages HTML.

### 1.9 Critères de succès

Le projet est réussi lorsque :

- le mandat de la phase 0 est tenu dans les poids, ou l'attente d'une donnée est écrite ;
- les données nécessaires sont au dossier, contrôlées, et leur origine est nommée ;
- les trois portefeuilles se comparent sur la même période et la même série de rendements ;
- le suivi, les tableaux de bord et le rapport se rouvrent à partir de ces fichiers.


## 2. Analyse des métiers

Les métiers étudiés sont Portfolio Analyst, Investment Data Analyst, Risk Analyst et Wealth Management Analyst. L'introduction nomme aussi l'Investment Operations Analyst. Il est traité sur le même plan, à la fin de cette section.

Les outils cités sous chaque métier sont ceux du poste dans une banque. Ce projet travaille avec les outils nommés au paragraphe 1.8.

### 2.1 Portfolio Analyst

Le travail exercé dans l'introduction : construire un portefeuille, le suivre, expliquer son résultat.

**Missions principales**

- Construire le portefeuille à l'intérieur des limites du mandat.
- Suivre l'allocation.
- Contrôler le respect des contraintes.
- Expliquer le résultat.
- Comparer le portefeuille au benchmark du mandat.
- Produire le rapport d'investissement.

**Outils utilisés**

Excel, Bloomberg, FactSet, Morningstar, Python, SQL.

**Indicateurs suivis**

Rendement, rendement annualisé, volatilité, ratio de Sharpe, ratio de Sortino, écart de suivi, ratio d'information, drawdown maximal.

**Compétences attendues**

Analyse financière, construction de portefeuille, statistiques, lecture d'un rapport.

**Dans ce projet**

La construction, le suivi et l'analyse de performance tiennent ce travail. La comparaison publiée est celle écrite dans `documentation/00_mandat/04_regles_des_portefeuilles.ipynb`.


### 2.2 Investment Data Analyst

Le travail exercé dans l'introduction : rassembler les données, en contrôler la qualité, en documenter l'origine.

**Missions principales**

- Rassembler les cours, le change et le volume.
- Réunir des sources distinctes dans les mêmes fichiers.
- Contrôler la qualité avant qu'un portefeuille les lise.
- Documenter l'origine de chaque série.
- Préparer les fichiers de la construction.

**Outils utilisés**

Python, SQL, interfaces de données de marché, bases de données, Excel.

**Indicateurs suivis**

Taux de complétude, taux de doublons, disponibilité des séries, contrôles de validité.

**Compétences attendues**

Python, SQL, qualité des données, rigueur de la source.

**Dans ce projet**

L'acquisition et le contrôle qualité tiennent ce travail. Une donnée manquante reste écrite comme manquante.


### 2.3 Risk Analyst

Le travail exercé dans l'introduction : mesurer le risque, lire les expositions, suivre les limites du mandat.

**Missions principales**

- Mesurer les pertes.
- Lire les expositions.
- Suivre les limites du mandat.
- Lire la concentration.
- Appliquer des chocs écrits à l'avance.

**Outils utilisés**

Python, SQL, Excel, Bloomberg.

**Indicateurs suivis**

Volatilité, drawdown maximal, concentration, exposition, value at risk, expected shortfall, résultat d'un test de résistance.

**Compétences attendues**

Finance quantitative, statistiques, lecture des limites, synthèse.

**Dans ce projet**

Le suivi tient les expositions, la concentration et les limites. La value at risk et l'expected shortfall sont les indicateurs du métier. Leur méthode est celle de l'analyse du risque et des tests de résistance.


### 2.4 Wealth Management Analyst

Le travail exercé dans l'introduction : traduire la situation du client en contraintes d'investissement.

**Missions principales**

- Lire le profil du client.
- Tenir les contraintes qui en découlent.
- Vérifier que le portefeuille reste dans ce profil.
- Préparer la lecture des résultats pour le client.

**Outils utilisés**

Excel, Bloomberg, outil de suivi de la relation client, outils de reporting.

**Indicateurs suivis**

Performance du portefeuille, allocation, diversification, exposition par secteur, par zone et par devise, niveau de risque accepté par le client.

**Compétences attendues**

Lecture d'un profil, allocation, clarté vis-à-vis du client.

**Dans ce projet**

Le profil et la politique d'investissement du 16 octobre 2026 tiennent ce travail. Le client accepte des pertes temporaires qui peuvent dépasser 30 %.


### 2.5 Investment Operations Analyst

Le travail exercé dans l'introduction : tenir les référentiels, et faire en sorte que chaque chiffre puisse être refait.

**Missions principales**

- Tenir l'identité de chaque titre.
- Tenir les devises et les séries qui convertissent un cours en francs.
- Contrôler qu'un fichier lu est bien le fichier reçu.
- Rendre un chiffre refaisable.

**Outils utilisés**

Python, SQL, listes de référence, contrôles de cohérence.

**Indicateurs suivis**

Unicité des identifiants, complétude des colonnes obligatoires, écart entre le fichier reçu et le fichier lu.

**Compétences attendues**

Contrôle, traçabilité, constance des identifiants.

**Dans ce projet**

L'identité des titres, les devises et le change sont tenus de façon qu'un rendement en francs puisse être refait.

## Dictionnaire

Ce dictionnaire reprend les termes employés dans cette note qui n'y sont pas déjà expliqués. Il n'en ajoute pas.

**Allocation.** Répartition du portefeuille entre des catégories, des secteurs, des zones ou des devises.

**Benchmark.** Point de comparaison du mandat. La référence stratégique est le MSCI ACWI Sustainable Development Index (Net). Net signifie : dividendes après impôt. Cette référence n'est pas achetée et n'a pas de cours propre. La comparaison publiée utilise le fonds iShares MSCI Global Sustainable Development Goals ETF, qui suit cet indice. La présence dans ce fonds n'est pas une preuve d'éligibilité ODD. L'absence n'est pas une preuve d'inéligibilité.

**Change.** Prix d'une monnaie contre une autre. Il sert ici à convertir un cours en francs.

**Concentration.** Fait qu'une part importante du portefeuille tienne dans peu de titres.

**Diversification.** Répartition qui évite de faire reposer le portefeuille sur un seul titre, un seul secteur ou une seule zone.

**Drawdown maximal.** Plus forte baisse depuis un plus haut déjà atteint.

**Écart de suivi.** Distance entre le rendement du portefeuille et celui du fonds nommé à Benchmark. Le plafond de 8 % lie les poids à partir d'octobre 2021.

**Expected shortfall.** Perte moyenne dans les cas situés au-delà du seuil de la value at risk.

**Exposition.** Part du portefeuille dans un secteur, une zone ou une devise.

**Ratio d'information.** Rendement au-dessus du fonds de comparaison, divisé par l'écart de suivi.

**Ratio de Sharpe.** Rendement au-dessus du taux sans risque, divisé par la volatilité.

**Ratio de Sortino.** Même lecture que le ratio de Sharpe. Le dénominateur ne retient que les variations défavorables.

**Référentiel.** Liste qui donne à chaque titre une identité stable.

**Rendement.** Variation du cours entre deux dates. Ici, d'un mois sur l'autre, en francs.

**Rendement annualisé.** Rendement ramené à une année.

**Taux de complétude.** Part des valeurs attendues qui sont présentes.

**Taux de doublons.** Part des lignes qui répètent une même identité.

**Test de résistance.** Choc écrit à l'avance, appliqué au portefeuille pour lire les pertes.

**Univers.** Liste des titres parmi lesquels les portefeuilles sont construits.

**Value at risk.** Perte associée à un seuil de la distribution des résultats.

**Volatilité.** Ampleur des variations du rendement.
