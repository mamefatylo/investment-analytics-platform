# 1. Présentation du projet

**Contexte**

Les banques privées, sociétés de gestion d'actifs et équipes de gestion des risques s'appuient quotidiennement sur des données financières afin de construire des portefeuilles, mesurer les risques, analyser les performances et accompagner la prise de décision d'investissement.

Ces données proviennent de multiples sources et doivent être :

- collectées ;
- contrôlées ;
- enrichies ;
- analysées ;
- transformées en indicateurs ;
- présentées sous forme de tableaux de bord et de rapports.

Le rôle d'un Data Analyst, d'un Investment Data Analyst ou d'un Portfolio Analyst consiste à transformer ces données en informations exploitables permettant aux décideurs de comprendre une situation et d'orienter leurs actions.

Dans ce contexte, ce projet vise à développer une plateforme complète de **Portfolio Construction, Monitoring & Risk Analytics** appliquée à un univers d'investissement aligné sur les Objectifs de Développement Durable (ODD) des Nations Unies.

L'objectif est de reproduire un environnement de travail proche de celui rencontré dans les métiers de :

- Portfolio Analyst ;
- Investment Data Analyst ;
- Risk Analyst ;
- Wealth Management Analyst ;
- Investment Operations Analyst.

**Vision globale du projet**

Le projet couvre l'ensemble de la chaîne de valeur d'une activité d'analyse d'investissement :

```text
Profil Investisseur
        ↓
Investment Policy Statement (IPS)
        ↓
Méthodologie d'Éligibilité ODD
        ↓
Construction de l'Univers Investissable
        ↓
Data Acquisition
        ↓
Data Quality
        ↓
Portfolio Construction
        ↓
Portfolio Monitoring
        ↓
Performance Analytics
        ↓
Risk Analytics
        ↓
Stress Testing
        ↓
Business Intelligence
        ↓
Reporting & Insights
```

# 2. Objectifs pédagogiques

À travers ce projet, l'objectif est de démontrer la capacité à reproduire des méthodes et pratiques utilisées dans les fonctions d'analyse d'investissement et de gestion de portefeuille.

---

**Comprendre un besoin métier**

Être capable de :

- identifier les attentes des utilisateurs ;
- comprendre les objectifs d'investissement ;
- traduire un besoin métier en besoin analytique ;
- définir des indicateurs pertinents ;
- concevoir une solution adaptée à un contexte financier.

---

**Maîtriser les données**

Être capable de :

- collecter les données ;
- contrôler leur qualité ;
- structurer les données ;
- documenter leur utilisation ;
- garantir leur traçabilité ;
- assurer leur reproductibilité.

---

**Produire des analyses**

Être capable de :

- étudier la performance d'un portefeuille ;
- mesurer les risques ;
- construire des scénarios ;
- identifier des tendances ;
- comparer différentes stratégies d'investissement ;
- produire des indicateurs d'aide à la décision.

---

**Communiquer efficacement**

Être capable de :

- construire des dashboards professionnels ;
- produire des rapports clairs ;
- formuler des recommandations ;
- présenter des résultats à un public non technique ;
- transformer des analyses complexes en informations compréhensibles.

# 3. Méthodologie de travail attendue

L'un des objectifs principaux du projet est de travailler selon une démarche proche de celle utilisée dans une entreprise de gestion d'actifs ou une banque privée.

Chaque phase du projet devra produire quatre éléments complémentaires.

---

**Une réflexion métier**

Avant toute implémentation technique, il est nécessaire de répondre à la question :

> Pourquoi cette étape existe-t-elle ?

L'objectif est de comprendre le besoin métier auquel répond chaque phase.

Cette réflexion doit permettre :

- d'identifier le problème à résoudre ;
- de comprendre les attentes des utilisateurs ;
- de justifier les travaux réalisés ;
- de relier les analyses aux métiers ciblés.

---

**Un livrable**

Chaque phase doit produire un livrable concret répondant à un besoin métier ou technique.

La question à se poser est :

> Qu'attendrait un manager, un client ou un utilisateur final ?

Selon les phases, les livrables pourront prendre la forme :

- d'un document d'analyse ;
- d'un schéma d'architecture ;
- d'une base de données ;
- d'un pipeline de traitement ;
- d'un portefeuille ;
- d'un tableau de bord ;
- d'un rapport de synthèse.

Le livrable doit être exploitable, compréhensible et directement utilisable.

---

**Une documentation**

Tout travail réalisé doit être documenté afin d'en garantir la compréhension et la reproductibilité.

La documentation doit permettre :

- de comprendre les objectifs ;
- de connaître les hypothèses retenues ;
- d'identifier les données utilisées ;
- de reproduire les traitements ;
- de justifier les décisions prises.

L'objectif n'est pas uniquement de produire un résultat mais également d'expliquer comment ce résultat a été obtenu.

---

**Une démonstration**

Chaque phase doit être accompagnée d'une preuve démontrant que le travail réalisé fonctionne correctement.

La question à se poser est :

> Comment démontrer que le résultat est valide ?

La démonstration peut prendre différentes formes :

- exemples de résultats ;
- extraits de données ;
- captures d'écran ;
- requêtes SQL ;
- notebooks Python ;
- visualisations ;
- tableaux de bord ;
- tests de validation.

Une phase est considérée comme terminée uniquement lorsque son fonctionnement peut être démontré.

---

**Métiers ciblés**

Le projet a été conçu pour reproduire les principales activités rencontrées dans plusieurs fonctions de l'investissement et de la gestion d'actifs.

| Métier | Contribution du projet |
|----------|----------|
| Portfolio Analyst | Construction et suivi de portefeuilles, analyse de performance, reporting d'investissement |
| Investment Data Analyst | Acquisition des données, contrôles qualité, structuration des référentiels et gouvernance des données |
| Risk Analyst | Mesure des risques, analyse des expositions, stress testing et suivi des limites |
| Wealth Management Analyst | Analyse des contraintes d'investissement, construction de l'univers investissable et reporting client |
| Investment Operations Analyst | Gestion des référentiels, qualité des données, traçabilité et auditabilité des traitements |

---

**Résultat attendu**

À l'issue du projet, l'objectif est de disposer d'une plateforme démontrant la capacité à transformer des données financières brutes en analyses exploitables pour :

- la construction de portefeuilles ;
- le suivi des investissements ;
- l'évaluation des performances ;
- la mesure des risques ;
- la production de reportings décisionnels ;
- l'aide à la décision dans un environnement de gestion d'actifs.

Le projet constitue un démonstrateur technique et métier destiné à mettre en évidence des compétences directement transférables vers les métiers de l'Investment Analytics, du Portfolio Analytics, du Risk Analytics et du Wealth Management.
