# Phase 2 – Architecture des données

Dans les métiers de l’investissement, de l’analyse financière et de la gestion des risques, la qualité des analyses dépend directement de la qualité des données utilisées.

Cette phase permet :

- d’identifier les sources de données utilisées par le projet ;
- de comprendre les flux de circulation des données ;
- de documenter les traitements appliqués aux données ;
- d’identifier les zones de stockage ;
- de garantir la traçabilité, la qualité et la reproductibilité des analyses.

L’objectif est de construire une architecture de données fiable permettant d’alimenter les analyses de portefeuille, de performance et de risque tout au long du projet.

```text
┌──────────────────────────────────────────┐
│            1. SOURCES EXTERNES           │
└──────────────────────────────────────────┘

• Données financières
• Données macroéconomiques
• Données ESG / ODD

                    ↓

┌──────────────────────────────────────────┐
│            2. SOURCE REGISTRY            │
└──────────────────────────────────────────┘

• Fournisseur
• URL
• Fréquence
• Méthode d'acquisition
• Niveau de fiabilité

                    ↓

┌──────────────────────────────────────────┐
│             3. ACQUISITION               │
└──────────────────────────────────────────┘

• API
• Export Excel
• Téléchargement CSV

                    ↓

┌──────────────────────────────────────────┐
│          4. ACQUISITION LOG              │
└──────────────────────────────────────────┘

• Date
• Fichier
• Source
• Statut
• Période couverte

                    ↓

┌──────────────────────────────────────────┐
│          5. RAW DATA LAYER               │
└──────────────────────────────────────────┘

Conservation des données brutes sans modification.

                    ↓

┌──────────────────────────────────────────┐
│         6. CONTRÔLES QUALITÉ             │
└──────────────────────────────────────────┘

• Identifiants
• Dates
• Doublons
• Cohérence
• Valeurs manquantes
• Couverture des données

                    ↓

┌──────────────────────────────────────────┐
│          7. STANDARDISATION              │
└──────────────────────────────────────────┘

• Formats
• Devises
• Identifiants
• Unités

                    ↓

┌──────────────────────────────────────────┐
│         8. SECURITIES MASTER             │
└──────────────────────────────────────────┘

Référentiel central
des instruments financiers

                    ↓

┌──────────────────────────────────────────┐
│         9. CANDIDATE UNIVERSE            │
└──────────────────────────────────────────┘

Construction de l'univers initial des titres candidats

                    ↓

┌──────────────────────────────────────────┐
│            10. IPS FILTERS               │
└──────────────────────────────────────────┘

Application des contraintes d'investissement

                    ↓

┌──────────────────────────────────────────┐
│            11. ODD FILTERS               │
└──────────────────────────────────────────┘

Application de la méthodologie d'éligibilité ODD

                    ↓

┌──────────────────────────────────────────┐
│          12. ELIGIBLE UNIVERSE           │
└──────────────────────────────────────────┘

Univers investissable final

                    ↓

┌──────────────────────────────────────────┐
│            13. SQL DATABASE              │
└──────────────────────────────────────────┘

Stockage structuré des données validées

                    ↓

┌──────────────────────────────────────────┐
│              14. ANALYTICS               │
└──────────────────────────────────────────┘

• Portfolio Construction
• Portfolio Monitoring
• Performance Analytics
• Risk Analytics
• Stress Testing

                    ↓

┌──────────────────────────────────────────┐
│              15. REPORTING               │
└──────────────────────────────────────────┘

• Dashboards Power BI
• Rapports
• Analyses d'investissement
```                  

## 2.3 Dictionnaire de données

Le dictionnaire de données documente les principales structures de données utilisées dans le projet. Pour chaque champ sont précisés :

- sa signification métier ;
- son rôle dans le processus de traitement ;
- son utilisation dans les analyses de portefeuille, de performance et de risque ;
- sa contribution à la gouvernance et à la traçabilité des données.

Cette documentation contribue à la compréhension, à la qualité et à la reproductibilité des analyses réalisées dans le projet.

### 2.3.1 Source Registry

| Champ | Description | Utilisation |
|---------|---------|---------|
| Source ID | Identifiant unique attribué à chaque fournisseur ou source de données utilisée dans le projet. | Permettre le suivi, le référencement et l'audit des sources utilisées tout au long du cycle de vie des données. |
| Provider | Organisme ou plateforme mettant les données à disposition (Yahoo Finance, FRED, SNB, OpenFIGI, etc.). | Identifier l'origine des données et évaluer leur niveau de confiance. |
| URL | Adresse web ou point d'accès permettant de récupérer les données. | Assurer la traçabilité et la reproductibilité des acquisitions de données. |
| Data / Service | Nature des informations fournies par la source (prix de marché, données fondamentales, données ESG, change, etc.). | Identifier les besoins couverts par chaque source et éviter les doublons. |
| Acquisition Method | Mode de récupération des données : API, téléchargement CSV, export Excel ou extraction manuelle. | Documenter le fonctionnement du pipeline d'acquisition et faciliter sa maintenance. |
| Reliability | Niveau de confiance accordé à la source sur la base de son caractère officiel, de sa réputation ou de sa qualité historique. | Prioriser les sources de référence et gérer les risques liés à la qualité des données. |
| Backup Source | Source alternative pouvant être utilisée en cas d'indisponibilité ou de problème avec la source principale. | Garantir la continuité du processus d'acquisition. |

### 2.3.2 Securities Master

| Champ | Description | Utilisation |
|---------|---------|---------|
| ISIN | International Securities Identification Number. Identifiant international unique d'un instrument financier. | Constituer la clé de référence principale permettant de relier l'ensemble des jeux de données du projet. |
| Ticker | Symbole boursier utilisé sur une place de cotation donnée. | Faciliter l'acquisition et l'identification des titres sur les marchés financiers. |
| Asset Name | Dénomination officielle de l'entreprise ou de l'instrument financier. | Améliorer la lisibilité des analyses et des reportings. |
| Exchange | Place boursière sur laquelle le titre est coté. | Identifier l'environnement de négociation et les contraintes de marché associées. |
| Country | Pays principal associé à l'émetteur ou à la cotation. | Réaliser des analyses géographiques et des contrôles réglementaires. |
| Region | Regroupement géographique plus large (Europe, Amérique du Nord, Asie-Pacifique, etc.). | Produire des analyses et reportings régionaux. |
| Currency | Devise de cotation du titre. | Gérer les conversions de change et les expositions monétaires. |
| Asset Type | Catégorie du titre (Action, ETF, etc.). | Différencier les instruments et appliquer des règles d'investissement adaptées. |
| Source | Source ayant fourni les informations de référence sur le titre. | Garantir la traçabilité des données de référence. |
| Source Date | Date de collecte ou de validation des informations. | Contrôler l'actualité et la fraîcheur des données. |

### 2.3.3 Market Dataset

| Champ | Description | Utilisation |
|---------|---------|---------|
| Date | Date de cotation correspondant à l'observation de marché. | Construire les séries temporelles utilisées dans les analyses. |
| Open | Premier prix observé à l'ouverture de la séance de cotation. | Réaliser des contrôles OHLC et certaines analyses techniques. |
| High | Prix le plus élevé observé durant la séance de cotation. | Vérifier la cohérence des données de marché. |
| Low | Prix le plus bas observé durant la séance de cotation. | Vérifier la cohérence des données de marché. |
| Close | Dernier prix observé à la clôture de la séance. | Servir de base à l'analyse des performances et aux calculs financiers. |
| Adjusted Close | Prix ajusté tenant compte des dividendes, splits et autres opérations sur titres. | Calculer des rendements historiquement cohérents. |
| Volume | Nombre total de titres échangés sur la séance. | Mesurer la liquidité et l'activité de marché d'un titre. |
| Dividends | Montant des dividendes distribués aux actionnaires. | Calculer la performance totale et analyser la politique de distribution. |
| Splits | Opérations de division ou de regroupement d'actions. | Ajuster les séries historiques de prix. |
| Corporate Actions | Événements ayant un impact sur le titre (fusion, spin-off, regroupement, etc.). | Expliquer les variations de prix et maintenir la cohérence des historiques. |

### 2.3.4 Fundamentals Dataset

| Champ | Description | Utilisation |
|---------|---------|---------|
| Revenue | Chiffre d'affaires total généré par l'entreprise au cours d'une période donnée grâce à la vente de biens et services. | Mesurer la taille de l'entreprise, analyser sa croissance et comparer son activité à celle de ses concurrents. |
| EBIT | Earnings Before Interest and Taxes. Résultat opérationnel mesurant le bénéfice généré par l'activité avant prise en compte des charges financières et de l'impôt. | Évaluer la rentabilité opérationnelle de l'entreprise et comparer la performance de sociétés ayant des structures financières différentes. |
| EBITDA | Earnings Before Interest, Taxes, Depreciation and Amortization. Indicateur de rentabilité opérationnelle mesurant la performance de l'entreprise avant intérêts, impôts et amortissements. | Comparer l'efficacité opérationnelle des sociétés et analyser leur capacité à générer des flux financiers issus de leur activité. |
| Net Income | Résultat net final de l'entreprise après déduction de l'ensemble des charges, intérêts, impôts et éléments exceptionnels. | Mesurer la profitabilité globale de l'entreprise et sa capacité à créer de la valeur pour les actionnaires. |
| Total Assets | Valeur totale des actifs détenus par l'entreprise (trésorerie, immobilisations, créances, stocks, etc.). | Évaluer la taille économique de l'entreprise et analyser sa structure financière. |
| Total Liabilities | Ensemble des obligations financières et dettes de l'entreprise envers des tiers. | Analyser le niveau d'endettement et les engagements financiers de l'entreprise. |
| Equity | Valeur résiduelle revenant aux actionnaires après déduction de l'ensemble des passifs de l'entreprise. | Évaluer la solidité financière de l'entreprise et sa capacité à absorber d'éventuelles pertes. |
| Total Debt | Montant total des dettes financières à court et long terme contractées par l'entreprise auprès de créanciers, banques ou investisseurs. | Mesurer le levier financier et évaluer le risque lié à l'endettement. |
| Cash | Trésorerie et équivalents de trésorerie immédiatement mobilisables par l'entreprise pour financer ses activités ou faire face à ses obligations financières. | Analyser la liquidité de l'entreprise et sa capacité à faire face à ses obligations à court terme. |
| FCF (Free Cash Flow) | Flux de trésorerie disponible après financement des investissements nécessaires au maintien et au développement de l'activité. | Évaluer la capacité de l'entreprise à générer du cash, financer sa croissance, verser des dividendes ou rembourser ses dettes. |
| Shares Outstanding | Nombre total d'actions en circulation détenues par les investisseurs. | Calculer la capitalisation boursière, les bénéfices par action (EPS) et d'autres ratios financiers. |

### 2.3.5 FX Dataset

| Champ | Description | Utilisation |
|---------|---------|---------|
| Date | Date d'observation du taux de change. | Synchroniser les conversions avec les données de marché. |
| Currency | Devise étrangère observée par rapport au CHF. | Identifier la monnaie d'origine des actifs. |
| Exchange Rate to CHF | Taux de change permettant de convertir la devise locale vers la devise de reporting (CHF). | Uniformiser les calculs de performance et de risque dans une devise unique. |

### 2.3.6 ODD Assessment

| Champ | Description | Utilisation |
|---------|---------|---------|
| ISIN | Identifiant unique du titre évalué. | Relier l'évaluation ODD aux autres référentiels du projet. |
| Company | Nom de la société analysée. | Faciliter la lecture des évaluations et des reportings. |
| SDG | Objectif de Développement Durable concerné par l'évaluation. | Classer les entreprises selon leur contribution potentielle aux ODD. |
| Activity | Activité économique soutenant l'analyse d'éligibilité. | Justifier le lien entre l'entreprise et l'ODD considéré. |
| Positive Contribution | Activité ou produit présentant une contribution identifiable à un ODD. | Évaluer les éléments favorables à l'éligibilité. |
| Negative Contribution | Activité susceptible de nuire à un ODD ou de contrevenir aux principes méthodologiques définis. | Identifier les risques de non-éligibilité. |
| Evidence | Élément documentaire utilisé pour justifier la décision (rapport annuel, publication officielle, etc.). | Assurer la transparence et l'auditabilité du processus. |
| Source | Origine des informations utilisées dans l'évaluation. | Garantir la traçabilité de la classification ODD. |
| Controversy | Événement ou problématique pouvant remettre en question l'éligibilité de l'entreprise. | Renforcer l'analyse de risque et le principe de non-préjudice. |
| Data Quality | Évaluation du niveau de confiance accordé aux informations collectées. | Mesurer la robustesse de la décision prise. |
| Status | Résultat final de l'évaluation (Eligible, Not Eligible, Indeterminate ou Excluded). | Alimenter la construction de l'univers investissable. |

### 2.3.7 Exclusion Register

| Champ | Description | Utilisation |
|---------|---------|---------|
| ISIN | Identifiant du titre concerné par l'exclusion. | Assurer une identification unique de l'instrument exclu. |
| Company | Nom de l'entreprise concernée. | Faciliter la lecture du registre. |
| Exclusion Reason | Raison principale ayant conduit à l'exclusion. | Justifier la décision prise. |
| Rule Applied | Référence à la contrainte IPS ou à la méthodologie ODD appliquée. | Garantir la cohérence des décisions d'exclusion. |
| Source | Origine des informations ayant conduit à la décision. | Assurer la traçabilité des exclusions. |
| Date | Date de l'exclusion ou de la décision. | Historiser les changements de statut. |
| Comments | Informations complémentaires sur la décision ou son contexte. | Documenter les cas particuliers et faciliter les revues futures. |

### 2.3.8 Audit Trail

| Champ | Description | Utilisation |
|---------|---------|---------|
| Audit ID | Identifiant unique de l'événement enregistré. | Permettre le suivi des opérations effectuées sur les données. |
| Date | Date et heure de l'action réalisée. | Reconstituer chronologiquement les transformations. |
| Action | Opération effectuée (création, modification, suppression, validation, etc.). | Comprendre la nature de l'intervention réalisée. |
| Dataset | Jeu de données concerné par l'action. | Identifier précisément la zone impactée. |
| Previous Value | Valeur enregistrée avant modification. | Assurer la traçabilité des changements. |
| New Value | Valeur enregistrée après modification. | Vérifier l'impact de la transformation. |
| Reason | Justification métier ou technique du changement. | Garantir la gouvernance et la transparence des traitements. |
| Source | Origine de l'information ayant déclenché la modification. | Faciliter les audits et contrôles qualité. |
| Owner | Personne ou processus ayant réalisé l'action. | Assurer l'attribution et la responsabilité des modifications. |
